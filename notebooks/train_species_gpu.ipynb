# MeatVision — Species model training on free Colab GPU

This notebook runs `Scripts/models/train_species.py` using the project code.

It is designed for the case where your dataset is available in Google Drive under `MyDrive/Dataset/` (no need to upload the whole ~43GB project to Drive).

**Expected dataset structure in Drive:**
- `MyDrive/Dataset/train/species/*`
- `MyDrive/Dataset/valid/species/*`
- `MyDrive/Dataset/test/species/*`


```python
# ========= Setup: GPU + Drive =========
from google.colab import drive
drive.mount('/content/drive')

# Check GPU
import torch
print('CUDA available:', torch.cuda.is_available())
print('GPU:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU')
```

```python
# ========= Copy/prepare project code =========
# This notebook assumes the repository is already present in the current Colab environment
# (e.g., you can upload this repo zip to Colab, or clone it).

# If you already have the code in the current runtime, skip this cell.

import os, sys
from pathlib import Path

# Detect a likely project root: if you uploaded the repo to Colab, it will be in /content/<repo>
possible_roots = ['/content', '/content/MeatVision_Project']
project_root = None
for r in possible_roots:
    p = Path(r)
    if (p / 'Scripts' / 'models' / 'train_species.py').exists():
        project_root = p
        break

if project_root is None:
    raise RuntimeError('Could not find Scripts/models/train_species.py. Upload/clone the repo so it is available under /content.')

print('Using project_root =', project_root)

# Ensure imports like `from config import *` work
models_dir = project_root / 'Scripts' / 'models'
sys.path.insert(0, str(models_dir))

# Also ensure project root is importable for other scripts
sys.path.insert(0, str(project_root))

print('Added to sys.path:', models_dir)
```

```python
# ========= Dataset path wiring =========
# Update this if your Drive folder name differs.
DRIVE_DATASET_ROOT = Path('/content/drive/MyDrive/Dataset')

# Training code expects: project_root/Dataset/{train,valid,test}/{species,freshness}
# We'll create the project_root/Dataset and copy/symlink the species folders.

expected_species = ['train/species', 'valid/species', 'test/species']
for rel in expected_species:
    src = DRIVE_DATASET_ROOT / rel
    if not src.exists():
        raise FileNotFoundError(f'Missing in Drive: {src}')

# Create project dataset directories
project_dataset_root = project_root / 'Dataset'
(project_dataset_root / 'train').mkdir(parents=True, exist_ok=True)
(project_dataset_root / 'valid').mkdir(parents=True, exist_ok=True)
(project_dataset_root / 'test').mkdir(parents=True, exist_ok=True)

# Species folders
for split in ['train', 'valid', 'test']:
    (project_dataset_root / split / 'species').mkdir(parents=True, exist_ok=True)

print('Dataset wiring OK. Copying species datasets...')

# Copy (simple + reliable). If you prefer symlinks, you can modify this.
import shutil
for rel in expected_species:
    src = DRIVE_DATASET_ROOT / rel
    dst = project_dataset_root / rel
    # Use copytree with dirs_exist_ok (Py3.10+) ; Colab typically has this
    shutil.copytree(src, dst, dirs_exist_ok=True)

print('Dataset copied into:', project_dataset_root)
```
